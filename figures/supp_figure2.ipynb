# Supplementary Figure 2

Behavioral data from the memory-guided visual search task (session 2) separated by target-competitor distance. Accompanies Figure 2. Reads `dataframes/search_data_all.csv`, produced by `figure2.ipynb` -- run that notebook first.

```python
import os.path as op

import pandas as pd

from mgvs_release.plotting import (
    setup_style, compute_resp_by_dist, plot_dist_error_hist, plot_dist_resp_swarm,
    plot_dist180_error_hist, plot_dist180_resp_swarm,
)
from mgvs_release.stats import print_dist_stats, paired_ttest
```

```python
setup_style()
```

```python
save_dir = '../plots/'
dataframes_dir = '../dataframes/'
data = pd.read_csv(op.join(dataframes_dir, 'search_data_all.csv'))
```

## Supplementary Fig. 2A: Distributions by target-competitor distance (first saccade)

```python
plot_dist_error_hist(data, 'sacc_first', 'First saccade frequency', 200,
                      op.join(save_dir, 'sfig2a_45-90-135.svg'))
```

```python
plot_dist180_error_hist(data, 'sacc_first', 3, 220, 'First saccade frequency',
                         op.join(save_dir, 'sfig2a_180.svg'))
```

## Supplementary Fig. 2B: Probabilities by target-competitor distance (first saccade)

```python
resp_by_dist_long = compute_resp_by_dist(data, 'sacc_first')
```

```python
plot_dist_resp_swarm(resp_by_dist_long, 'First saccade probability',
                      op.join(save_dir, 'sfig2b_45-90-135.svg'))
```

```python
print_dist_stats(resp_by_dist_long)
```

```python
plot_dist180_resp_swarm(resp_by_dist_long, 'First saccade probability',
                         op.join(save_dir, 'sfig2b_180.svg'))
```

```python
# 180 degree pairs: target vs competitor saccades 
paired_ttest(resp_by_dist_long.query("targ_comp_dist==180 & resp_type=='targ_resp'")['prob'],
             resp_by_dist_long.query("targ_comp_dist==180 & resp_type=='comp_resp'")['prob'])
```

## Supplementary Fig. 2C: Distributions by target-competitor distance (last saccade)

```python
plot_dist_error_hist(data, 'sacc_last', 'Last saccade frequency', 205,
                      op.join(save_dir, 'sfig2c_45-90-135.svg'))
```

```python
plot_dist180_error_hist(data, 'sacc_last', 1, 205, 'Last saccade frequency',
                         op.join(save_dir, 'sfig2c_180.svg'))
```

## Supplementary Fig. 2D: Probabilities by target-competitor distance (last saccade)

```python
resp_by_dist_long = compute_resp_by_dist(data, 'sacc_last')
```

```python
plot_dist_resp_swarm(resp_by_dist_long, 'Last saccade probability',
                      op.join(save_dir, 'sfig2d_45-90-135.svg'))
```

```python
print_dist_stats(resp_by_dist_long)
```

```python
plot_dist180_resp_swarm(resp_by_dist_long, 'Last saccade probability',
                         op.join(save_dir, 'sfig2d_180.svg'))
```
