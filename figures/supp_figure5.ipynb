# Supplementary Figure 5

Preparatory coding in early visual cortex as a function of behavioral outcome defined by the last and first saccade of search. Accompanies Figure 4. Reads in `dataframes/evc_prep_curve_last.csv` and `dataframes/evc_prep_curve_first.csv` computed by `precompute/fig4_glm_resids_aggregate.py`

```python
import os.path as op

import pandas as pd

from mgvs_release.plotting import setup_style, plot_behav_response_panels

setup_style()

save_dir = "../plots/"
```

## Supplementary Fig. 5A: Mean EVC response with individual data points by behavioral outcome (last saccade)

```python
search_data_all = pd.read_csv(op.join('../dataframes', 'evc_prep_curve_last.csv'))
```

```python
search_data_all['resp_type'] = 'other'
search_data_all.loc[search_data_all['next_warp_resp']==True, 'resp_type'] = 'warp'
search_data_all.loc[search_data_all['next_comp_resp']==True, 'resp_type'] = 'competitor'

search_data_mean = search_data_all.groupby(
    ['subj', 'cue_validity', 'cond', 'resp_type']).mean(numeric_only=True).reset_index()
```

```python
plot_behav_response_panels(search_data_mean, op.join(save_dir, 'sfig5a.svg'))
```

## Supplementary Fig. 5B: Mean EVC response with individual data points by behavioral outcome (first saccade)

```python
search_data_all = pd.read_csv(op.join('../dataframes', 'evc_prep_curve_first.csv'))
```

```python
search_data_all['resp_type'] = 'other'
search_data_all.loc[search_data_all['next_warp_resp']==True, 'resp_type'] = 'warp'
search_data_all.loc[search_data_all['next_comp_resp']==True, 'resp_type'] = 'competitor'


search_data_mean = search_data_all.groupby(
    ['subj', 'cue_validity', 'cond', 'resp_type']).mean(numeric_only=True).reset_index()
```

```python
plot_behav_response_panels(search_data_mean, op.join(save_dir, 'sfig5b.svg'))
```
