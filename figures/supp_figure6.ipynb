# Supplementary Figure 6

Brain regions outside the hippocampus are not associated with differences between target and competitor activation in visual cortex. Accompanies Figure 5. Reads in `dataframes/*_rank_corr.csv` and `dataframes/*_rank_univ.csv` computed by `precompute/fig5_glm_resids_aggregate.py`.

```python
import os.path as op

import pandas as pd

from mgvs_release.plotting import setup_style, plot_roi_rank_swarm

setup_style()

save_dir = "../plots/"
```

## Supplementary Fig. 6A: Replotting HIPP with individual data points

```python
hipp_df = pd.read_csv(op.join("../dataframes", "hipp_rank_corr.csv"))
```

```python
hipp_mean = hipp_df.query(
    "cue_validity=='valid' & next_targ_comp_dist==45"
).groupby(['subj', 'rank_bin', 'cond'])[['bold', 'bold_z']].mean().reset_index()
```

```python
plot_roi_rank_swarm(hipp_mean, [-.2, .4],
                     'Less distinct\nhippocampal representations\n',
                     'More distinct\nhippocampal representations\n',
                     op.join(save_dir, 'sfig6a.svg'))
```
